# Business Recommendations — Google Play Store Analysis

These recommendations are drawn from the EDA findings (`Insights_and_Report.md`). They're
grouped by audience, since "what should I do with this data" looks different depending on
whether you're an app developer, a platform/store owner, or an investor screening categories.

> **Caveat before acting on any of this:** the category-level totals feeding these
> recommendations were affected by a duplicate-row bug in `Data_Cleaning.ipynb` (see README).
> The *direction* of each recommendation is unlikely to change once fixed, but exact figures
> (e.g. "35M installs per app") should be re-pulled from the corrected data before being used
> in anything client-facing.

## For App Developers

1. **Price paid apps in the $1–$5 range.** This band produced the most apps that were
   simultaneously above-average on both installs and rating (73 apps) — more than any other
   price tier. There's no evidence in this data that pricing above $5 buys more installs or
   higher ratings, so higher price points appear to only cut demand without a satisfaction
   trade-off in return.
2. **Don't rely on app size or update frequency as a differentiator.** Both showed
   negligible correlation with installs and rating. Engineering effort is better spent
   elsewhere than optimizing app size or update cadence purely for discoverability.
3. **Target underserved-but-proven categories rather than the most crowded ones.** Family
   and Game have the most competing apps; Communication has strong demand relative to its
   (smaller) competitor count. A new entrant is statistically more likely to find room in a
   category with high demand and moderate competition than in the most saturated ones.
4. **Consider Tools or Video Players if you can execute well.** Both categories have strong
   installs but the lowest average ratings (~4.08) among high-reach categories — a real
   quality gap exists between demand and user satisfaction there.
5. **If building a paid product, Education is the strongest category on record** — highest
   combination of average installs (~150K) and average rating (~4.75) among paid apps.

## For the Platform / Store Owner

1. **Investigate the Tools and Video Players categories for a quality problem, not just a
   competition problem.** High installs with below-average ratings across the board suggests
   users are downloading these apps out of necessity (utility, function) rather than
   satisfaction — an opportunity to improve curation, ranking signals, or developer guidelines
   specifically in these categories.
2. **Rating alone is a weak signal for surfacing "good" apps.** Since rating barely
   correlates with installs and reviews barely correlate with rating, install-count-based or
   review-count-based ranking may already be doing most of the work rating is meant to do —
   worth reviewing whether rating should carry more or less weight in discovery/ranking
   algorithms.
3. **The free/paid split (92%/8%) is stable and heavily favors free.** Store policies aimed
   at growing paid-app adoption would be working against a strong existing pattern — any such
   push should be tested carefully rather than assumed to shift behavior easily.

## For Investors / Market Analysts

1. **Use "installs per app" (demand relative to competition), not raw category size, to
   screen categories.** Family has the most apps but that's a crowded-market signal, not an
   opportunity signal. Communication's high installs-per-app ratio is a better opportunity
   metric, with the caveat that it may reflect a few dominant incumbents rather than open
   white space — worth a follow-up look at the underlying apps before treating it as
   "underserved."
2. **Content rating skew ("Everyone" at 80%) indicates the store's addressable audience is
   still mostly general/family audiences** rather than niche or restricted content — relevant
   context for any market-sizing exercise on this data.
