# Business  question

```python
relationship=playstoredata.groupby("Category").agg(
    {
        "Installs":"sum",
        "Rating":"mean"
    })
averagesuminstalls=relationship["Installs"].mean()
averagemeanrating=relationship["Rating"].mean()
#now
result=relationship[
    (relationship["Installs"]>averagesuminstalls)&
    (relationship["Rating"]<averagemeanrating)
]
print(result)
# visualizing
plt.figure(figsize=(12,6))
plt.scatter(relationship["Installs"],
            relationship["Rating"])
for category, row in result.iterrows():
    plt.annotate(
        category,
        (row["Installs"], row["Rating"]),
        xytext=(5,5),
        textcoords="offset points"
    )
plt.xscale("log")

plt.title("Sum of installs vs average rating of each category")
plt.xlabel("Sum of category Installs")
plt.ylabel("Average category rating")
plt.show()
```

# Business  question
- Which category has most successfull paid apps

```python
relationship=pd.pivot_table(
    data=playstoredata,
    index="Category",
    columns="Type",
    values=["Installs","Rating"],
    aggfunc="mean"
)
installthershold=relationship["Installs"].mean()
ratingthershold=relationship["Rating"].mean()
install_threshold = relationship[("Installs", "Paid")].mean()
rating_threshold = relationship[("Rating", "Paid")].mean()

result = relationship[
    (relationship[("Installs", "Paid")] > install_threshold) &
    (relationship[("Rating", "Paid")] > rating_threshold)
]
print(result)
# visualizing
plt.figure(figsize=(12,6))
scatter=plt.scatter(result[("Installs","Paid")],
            result[("Rating","Paid")])
for category,rows in result.iterrows():
    plt.annotate(category,
                 (rows[("Installs","Paid")],
                  rows[("Rating","Paid")]),
                  xytext=(5,5),
                  textcoords="offset points"
                  )
plt.title("Relationship b/w category and most sucessfull app")
plt.xlabel(" Paid app  average Installs ")
plt.ylabel("Averge rating")
plt.show()
```

# Business question
- which price range performs best for paid apps

```python
paidapps=playstoredata[playstoredata["Type"]=="Paid"]
ratingthershold=paidapps["Rating"].mean()
installsthreshold=paidapps["Installs"].mean()
successfullpaidapps=paidapps[
    (paidapps["Installs"]>installsthreshold ) & 
    (paidapps["Rating"]>ratingthershold)]
print(paidapps["Price"].max())
# now
bins=[0,1,5,10,20,50,100,200,300,400]
labels= [
    "$0-1",
    "$1–5",
    "$5–10",
    "$10–20",
    "$20–50",
    "$50–100",
    "$100-200",
    "$200-300",
    "$300+"
]
successfullpaidapps["Pricerange"]=pd.cut(
    successfullpaidapps["Price"],
    bins=bins,
    labels=labels
)
result=successfullpaidapps["Pricerange"].value_counts().sort_values()
print(result)
#visualizing
plt.figure(figsize=(10,8))
bar=plt.bar(result.index,result.values,edgecolor="black",color="blue")
plt.title("Relationship b/w price range and successfull paid apps")
plt.xlabel("Price Range")
plt.ylabel("Count of apps in particular price range")
plt.bar_label(bar,"%.0f")
plt.show()
```

# Answer
- $1-5 price range of paid apps have most average rating and average installs (73 succfull paid apps)

```python
# Calculate how many days ago each app was updated
latest_date = playstoredata["Last Updated"].max()
playstoredata["Days Since Update"] = (
    latest_date - playstoredata["Last Updated"]
).dt.days
# Check correlation
correlation = playstoredata["Days Since Update"].corr(
    playstoredata["Installs"]
)
print("Correlation:", correlation)
#visualization
plt.figure(figsize=(12, 6))

plt.scatter(
    playstoredata["Days Since Update"],
    playstoredata["Installs"],
    alpha=0.5
)

plt.xlabel("Days Since Last Update")
plt.ylabel("Installs")
plt.title("Relationship Between Update Recency and Installs")
plt.savefig("../IMAGES/updaterecencyvsinstalls.png")
plt.show()
```

# Answer 
-Recently updated apps tend to have slightly more installs, but the relationship is very weak (correlation = -0.1009). Therefore, update recency does not appear to be a strong factor influencing installs.

# Business question
- Which category has high demand but low compitition

```python
data = playstoredata.drop_duplicates("App")

# Group by category
relationship = data.groupby("Category").agg(
    installs=("Installs", "sum"),
    apps=("App", "count")
)

# Calculate demand relative to competition
relationship["installs_per_app"] = (
    relationship["installs"] / relationship["apps"]
)

# Sort categories by installs per app
relationship = relationship.sort_values(
    "installs_per_app",
    ascending=False
)

print(relationship)

# Visualization
plt.figure(figsize=(12, 8))

plt.scatter(
    relationship["apps"],
    relationship["installs_per_app"]
)

plt.xscale("log")
plt.yscale("log")

plt.title("Demand vs Competition by App Category")
plt.xlabel("Number of Apps (Competition)")
plt.ylabel("Average Installs per App (Demand)")
plt.savefig("../IMAGES/highdemandlowcompitition.png")
plt.show()
```

# Answer
- Communication is a high-demand, relatively low-competition category. It has approximately 11.04 billion installs across only 315 apps, giving it around 35 million installs per app. Compared with other high-demand categories such as Games and Tools, Communication has fewer competing apps while maintaining very high demand.

#  Business question
- Which category has high installs and high ratings

```python
relationship=playstoredata.groupby("Category").agg(
    installs=("Installs","sum"),
    rating=("Rating","mean")
)
installsthreshold=relationship["installs"].median()
ratingthershold=relationship["rating"].median()
result=relationship[(relationship["installs"]>installsthreshold)&(relationship["rating"]>ratingthershold)]
result=result.sort_values(["installs","rating"],ascending=[False,False])
print(result)
#visualizations
plt.figure(figsize=(12,6))
plt.scatter(relationship["rating"],relationship["installs"])
plt.title("relationship b/w Average rating and sum of installs")
plt.xlabel("Average rating")
plt.ylabel("Sum of Installs")
plt.savefig("../IMAGES/Categorywithhighinstallsandhighrating.png")
plt.show()
```

# Answer
- Game ,Productivity and Social categories are the categories with high rating and high installs
