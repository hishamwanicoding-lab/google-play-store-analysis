```python
#importing liabraries
import numpy as np
import pandas as pd 
import csv 
#reading files 
playstoredata=pd.read_csv("../DATA/RawData/googleplaystore.csv")
processedpscopy=playstoredata.copy()
processedplaystoredata=pd.read_csv("../DATA/RawData/ProcessedData/processedplaystoredata.csv")
```

handling playstoredata dublicated rows

```python
duplicates = playstoredata[playstoredata.duplicated()]
updatedplatstoredata=playstoredata.drop_duplicates()

processedpscopy=updatedplatstoredata
```

checking index sequence


```python
print(playstoredata.index)
```

removing completely empty rows and columns

```python
updatedplatstoredata=playstoredata.dropna(how='all')
processedpscopy=updatedplatstoredata
```

CLEANING APPS COLUMN


```python
apps=playstoredata["App"]
#finding missing values 
print(apps.isna().sum())
#checking uniques
# print(apps.nunique())
updatedapps=apps.str.strip()
processedpscopy["App"]=updatedapps
```

CLEANING CATOGORY



```python
catogory=playstoredata["Category"]
#checking unique values
# print(catogory.unique())
#row for invalid catogory value
# print(playstoredata[catogory=="1.9"])
updatedcatogory=catogory.str.replace("1.9","LIFESTYLE")
updatedcatogory=updatedcatogory.str.strip()
#strping text
processedpscopy["Category"]=updatedcatogory

```

CLEANING RATING FILE

```python
#
rating=playstoredata["Rating"]
#uniques
# print(updatedrating.unique())
# print(updatedrating.info())
updatedrating=rating.replace(19.0,np.nan)
# print(rating.isna().sum())
updatedrating=updatedrating.fillna(rating.median())
processedpscopy["Rating"]=updatedrating

#
```

cleaning review file 

```python
review=playstoredata["Reviews"]
# print(review.unique())
print(playstoredata[review=="3.0M"])
#removing character
updatedreview=review.replace("3.0M",np.nan)
# print(updatedreview.isna().sum())
updatedreview=updatedreview.astype(float)
# print(updatedreview.dtype)
updatedreview=updatedreview.fillna(updatedreview.median())
processedpscopy["Reviews"]=updatedreview
```

size coloumn cleaning

```python
size = playstoredata["Size"].str.replace(",", "", regex=False)
size = size.replace("Varies with device", np.nan)

# Create a copy to clean
updatedsize = size.copy()

mask_m = updatedsize.str.endswith("M", na=False)
mask_k = updatedsize.str.endswith("k", na=False)

updatedsize.loc[mask_m] = (
    updatedsize[mask_m]
    .str.replace("M", "", regex=False)
    .astype(float)
)

updatedsize.loc[mask_k] = (
    updatedsize[mask_k]
    .str.replace("k", "", regex=False)
    .astype(float) / 1000
)

# Convert entire column to numeric
updatedsize = pd.to_numeric(updatedsize, errors="coerce")

# Fill missing values with the median
updatedsize = updatedsize.fillna(updatedsize.median())

# Assign to your copied dataframe
processedpscopy["Size"] = updatedsize
```

cleaning installs

```python
installs=playstoredata["Installs"]
# print(installs.unique())
# print(installs.nunique())
#removing unessery characters
updattedinstalls=installs.str.replace(",","").str.replace("+","").replace("Free",np.nan)
#converting data type
updattedinstalls=updattedinstalls.astype(float)
updattedinstalls=updattedinstalls.fillna(updattedinstalls.median())
processedpscopy["Installs"]=updattedinstalls
```

CLEANING TYPE COLOUMN


```python
_type=playstoredata["Type"]
# print(_type.nunique())
# print(_type.unique())
#finding column
print(playstoredata[_type=="0"])
# updating
updatedtype=_type.str.replace("0","Free")
#filling nan 
updatedtype=updatedtype.fillna(updatedtype.mode()[0])
processedpscopy["Type"]=updatedtype
# print(_type.unique())
```

cleaning price column


```python
price=playstoredata["Price"]
# print(price.unique())
# print(playstoredata[price=="Everyone"])
#updating
updatedprice=price.str.replace("Everyone","0").str.replace("$","").astype(float)

processedpscopy["Price"]=updatedprice
```

cleaning content rating column

```python
contentrating=playstoredata["Content Rating"]
# print(contentrating.isna().sum())
# print(playstoredata[contentrating.isna()])
# print(contentrating.unique())
updatedcontentrating=contentrating.fillna("Everyone")
# print(updatedcontentrating.isna().sum())
processedpscopy["Content Rating"]=updatedcontentrating
```

cleaning generes column


```python
generes=playstoredata["Genres"]
# print(generes.isna().sum())
# print(generes.nunique())    
updatedgeneres=generes.str.strip()
processedpscopy["Genres"]=updatedgeneres
```

cleaning last updated column



```python
#cleaning last updated
lastupdated=pd.to_datetime(playstoredata["Last Updated"],format="mixed",errors="coerce")
# print(playstoredata[lastupdated.isna()])
updatedlastupdated=lastupdated.fillna(pd.Timestamp("january,1,2019"))
# print(updatedlastupdated.unique())
processedpscopy["Last Updated"]=updatedlastupdated
print(processedpscopy["Last Updated"].dtype)
```

cleaning current version column

```python
currentversion=playstoredata["Current Ver"]
# print(currenversion.isna().sum())
# print(playstoredata[currenversion.isna()])
updatedcurrentversion=currentversion.fillna("Unknown")
# print(updatedcurrentversion.nunique())
# print(updatedcurrentversion.unique())
processedpscopy["Current Ver"]=updatedcurrentversion
```

cleaning android version column

```python
androidversion=playstoredata["Android Ver"]
# print(androidversion.isna().sum())
updatedandroidversion=androidversion.fillna("Unknown")
updatedandroidversion=updatedandroidversion.str.strip()
# print(updatedandroidversion.unique())
processedpscopy["Android Ver"]=updatedandroidversion
```

updatingcolumn names


```python
columns=playstoredata.columns
updatedcolumns=playstoredata.columns.str.replace("Android ver","Android Version Required").str.replace("Current Ver","Current Version")
processedpscopy.columns=updatedcolumns

```
